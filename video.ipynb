```python
import cv2
import os
import cv2.colored_kinfu
import numpy as np
from tqdm import tqdm

def load_images_from_directory(directory):
    # Get list of files and sort them lexicographically
    files = sorted([os.path.join(directory, f) for f in os.listdir(directory) if f.endswith(('png', 'jpg', 'jpeg'))])
    # Load images
    images = [cv2.imread(file) for file in files]
    return images

def create_video(dir, output_video, coords, bbox, fps=30):

    images = load_images_from_directory(dir)
    height, width, _ = images[0][bbox[0]:bbox[1],bbox[2]:bbox[3],:].shape

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    video = cv2.VideoWriter(output_video, fourcc, fps, (width, height))
    
    for i, image in enumerate(tqdm(images)):
        image = image[bbox[0]:bbox[1],bbox[2]:bbox[3],:]
        for coord in coords[i]:
            image[int(coord[0]),int(coord[1])] = [0,0,255]

        video.write(image)
    
    video.release()
    print(f"Video saved as {output_video}")

def read_coordinates(file_path):
    frames = {}
    bbox = None
    with open(file_path, 'r') as f:
        for line in f:
            parts = line.strip().split('\t')
            if len(parts) == 1:
                bbox = line[line.find("(")+1:line.find(")")]
                bbox = tuple(map(int, bbox.split(', ')))
            if len(parts) == 5:
                frame_id = int(parts[0])
                x = float(parts[2])
                y = float(parts[3])
                
                if frame_id not in frames:
                    frames[frame_id] = []
                frames[frame_id].append((x, y))
    return frames, bbox

```

```python
dir = '/home/yuming/Documents/mt_data/MT10_30min_200x_1500_138_146pm'
output_video = '/home/yuming/Documents/dev/python/projects/microtuble-tracking/output_video_4.mp4'
coords_file = '/home/yuming/Documents/dev/python/projects/microtubule-tracking-2/rough_coordinates.txt'
coords, bbox = read_coordinates(coords_file)


```

```python
create_video(dir, output_video, coords, bbox)
```
